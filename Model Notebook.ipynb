```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## GET THE DATA

```python
df_train = pd.read_csv('train.csv', index_col=0)
df_test = pd.read_csv('test.csv', index_col=0)
```

## CONCATING BOTH DF INTO ONE

```python
df = pd.concat([df_train, df_test], axis=0)
```
